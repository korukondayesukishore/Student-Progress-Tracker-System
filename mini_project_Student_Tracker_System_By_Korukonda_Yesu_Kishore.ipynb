# Welcome Screen

```python

def welcome_screen():
    print("-----------------------------------------")
    print("      STUDENT PROGRESS TRACKER SYSTEM    ")
    print("-----------------------------------------")
```

# STUDENT INFORMATION

```python

def student_info():
    name = input("Enter Your Name: ")
    class_name = input("Enter Your Class: ")
    roll_number = int(input("Enter Your Roll Number: "))
    return name, class_name, roll_number
```

# Subject Marks

```python

def subject_marks():
    no_sub = int(input("Enter Number of Subjects: "))
    marks_list = []
    total_marks = 0

    for i in range(1, no_sub + 1):
        marks = int(input(f"Enter marks of subject {i}: "))
        marks_list.append(marks)
        total_marks += marks

    avrtg = total_marks / no_sub
    return marks_list, total_marks, avrtg
```

# Grade Calculation

```python

def grade_calculation(avrtg):
    if avrtg >= 90:
        return "A Grade"
    elif avrtg >= 75:
        return "B Grade"
    elif avrtg >= 50:
        return "C Grade"
    elif avrtg >= 35:
        return "D Grade"
    else:
        return "Fail"
```

# Pass / Fail

```python

def pass_fail_logic(marks_list):
    fail_count = sum(1 for m in marks_list if m < 35)
    result = "PASS" if fail_count == 0 else "FAIL"
    return fail_count, result
```

# Remarks

```python

def remarks_calculation(percentage):
    if percentage >= 90:
        return "Excellent"
    elif percentage >= 75:
        return "Good"
    elif percentage >= 50:
        return "Needs Improvement"
    else:
        return "Poor"
```

# Progress Report

```python

def student_report(name, class_name, roll_number, total_marks, avrtg, grade, result,
                   highest, lowest, percentage, remarks):
    print("-----------------------------------------")
    print("        STUDENT PROGRESS REPORT")
    print("-----------------------------------------")
    print(f"Name        : {name}")
    print(f"Class       : {class_name}")
    print(f"Roll No     : {roll_number}")
    print(f"Total Marks : {total_marks}")
    print(f"Average     : {avrtg}")
    print(f"Grade       : {grade}")
    print(f"Result      : {result}")
    print(f"Percentage  : {percentage:.2f}%")
    print(f"Highest     : {highest}")
    print(f"Lowest      : {lowest}")
    print(f"Remarks     : {remarks}")
    print("-----------------------------------------")
```

# Save report

```python
def save_report(name, class_name, roll_number, total_marks, avrtg, grade, result,
                highest, lowest, percentage, remarks):
    save = input("Do you want to save the report to a text file? (yes/no): ").lower()
    if save == "yes":
        filename = f"{name}_report.txt"
        with open(filename, "w") as file:
            file.write("-----------------------------------------\n")
            file.write("        STUDENT PROGRESS REPORT\n")
            file.write("-----------------------------------------\n")
            file.write(f"Name        : {name}\n")
            file.write(f"Class       : {class_name}\n")
            file.write(f"Roll No     : {roll_number}\n")
            file.write(f"Total Marks : {total_marks}\n")
            file.write(f"Average     : {avrtg}\n")
            file.write(f"Grade       : {grade}\n")
            file.write(f"Result      : {result}\n")
            file.write(f"Percentage  : {percentage:.2f}%\n")
            file.write(f"Highest     : {highest}\n")
            file.write(f"Lowest      : {lowest}\n")
            file.write(f"Remarks     : {remarks}\n")
            file.write("-----------------------------------------\n")
        print(f"Report saved successfully as {filename}")
```

# Menu System

```python

def run_system():
    welcome_screen()
    name, class_name, roll_number = student_info()

    marks_list, total_marks, avrtg = subject_marks()
    highest = max(marks_list)
    lowest = min(marks_list)
    percentage = (total_marks / (len(marks_list) * 100)) * 100

    grade = grade_calculation(avrtg)
    fail_count, result = pass_fail_logic(marks_list)
    remarks = remarks_calculation(percentage)

    student_report(name, class_name, roll_number, total_marks, avrtg, grade, result,
                   highest, lowest, percentage, remarks)

    while True:
        print("\n1. Enter marks again")
        print("2. View report again")
        print("3  Save the report to file")
        print("4. Exit to restart system")

        choice = input("Enter choice: ")

        if choice == "1":
            marks_list, total_marks, avrtg = subject_marks()
            highest = max(marks_list)
            lowest = min(marks_list)
            percentage = (total_marks / (len(marks_list) * 100)) * 100
            grade = grade_calculation(avrtg)
            fail_count, result = pass_fail_logic(marks_list)
            remarks = remarks_calculation(percentage)

        elif choice == "2":
            student_report(name, class_name, roll_number, total_marks, avrtg, grade, result,
                           highest, lowest, percentage, remarks)
        elif choice == "3":
          save_report(name, class_name, roll_number, total_marks, avrtg, grade, result,
                highest, lowest, percentage, remarks)

        elif choice == "4":
            break
        else:
            print("Invalid choice")
```

# Restart System

```python
def restart_system():
    while True:
        run_system()
        i=input("Do you want to restart the system? (yes/no): ").lower()
        if i != "yes":
            break
```
